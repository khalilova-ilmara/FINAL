```python
import pandas as pd

# Загрузка данных
file_path = "games.csv"
data = pd.read_csv(file_path)

# Просмотр структуры данных
print(data.info())
print(data.head())

# Обработка пропущенных значений
# Заполнение пропущенных значений медианой (для числовых данных)
data = data.fillna(data.median(numeric_only=True))

# Удаление дубликатов
data = data.drop_duplicates()

# Очистка текстовых данных (удаление лишних пробелов)
for col in data.select_dtypes(include=['object']):
    data[col] = data[col].str.strip()

# Преобразование временных меток (если есть столбцы с датами)
if 'date_column' in data.columns:  # Замените 'date_column' на название вашего столбца
    data['date_column'] = pd.to_datetime(data['date_column'])

# Нормализация числовых данных
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
numeric_cols = data.select_dtypes(include=['float64', 'int64']).columns
data[numeric_cols] = scaler.fit_transform(data[numeric_cols])

# Сохранение предобработанных данных
data.to_csv("games_preprocessed.csv", index=False)

print("Предобработка завершена. Данные сохранены в файл games_preprocessed.csv.")
```

```python
# Проверим, есть ли категориальные столбцы
categorical_columns = data.select_dtypes(include=['object']).columns
print("Категориальные столбцы:", categorical_columns)

if len(categorical_columns) > 0:
    for col in categorical_columns:
        print(f"\nАнализ столбца {col}:")
        print(data[col].value_counts())
else:
    print("Категориальные столбцы отсутствуют или их невозможно определить.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Распределение рейтингов
rating_counts = data['rating'].value_counts()

plt.figure(figsize=(10, 6))
sns.barplot(x=rating_counts.index, y=rating_counts.values, palette='viridis')
plt.title("Распределение рейтингов игр")
plt.xlabel("Рейтинг")
plt.ylabel("Количество")
plt.xticks(rotation=45)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Построим гистограммы для числовых столбцов
numeric_cols = data.select_dtypes(include=['float64', 'int64']).columns
data[numeric_cols].hist(figsize=(10, 8), bins=20)
plt.tight_layout()
plt.show()
```

```python
#4. Анализ временных данных
```

```python
# Проверка формата временных данных
if 'date_release' in data.columns:
    data['date_release'] = pd.to_datetime(data['date_release'], errors='coerce')
    print("\nПервые 20 значений столбца date_release:")
    print(data['date_release'].head(20))

    # Построим график количества релизов по годам
    data['release_year'] = data['date_release'].dt.year
    release_year_counts = data['release_year'].value_counts().sort_index()

    plt.figure(figsize=(10, 6))
    plt.plot(release_year_counts.index, release_year_counts.values, marker='o')
    plt.title("Количество релизов игр по годам")
    plt.xlabel("Годы")
    plt.ylabel("Количество релизов")
    plt.grid()
    plt.show()
```
